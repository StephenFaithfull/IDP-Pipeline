```python
import easyocr
reader = easyocr.Reader(['en']) # this needs to run only once to load the model into memory
```

```python
result = reader.readtext('Alice.JPG')
print(result)
```

```python
print(f"Text: {result[1]}, Coordinates: {result[0]}")
```

```python
raw_text = []
for res in result:
    print(f"Text: {res[1]}, Coordinates: {res[0]}")
    raw_text.append(res[1])
```
